```python
import tiktoken

# 1. Install: pip install tiktoken

# 2. Get an encoding for a specific model
# For gpt-4, gpt-3.5-turbo, etc.
encoding = tiktoken.encoding_for_model("gpt-4")
# For gpt-4o, gpt-4o-mini
# encoding = tiktoken.encoding_for_model("gpt-4o")

# 3. Encode text to tokens (list of integers)
text = "Hello, world! How are you?"
tokens = encoding.encode(text)
print(f"Tokens: {tokens}")
# Output (example): Tokens: [15339, 11, 18070, 281, 263, 12, 241, 286, 13]

# 4. Count tokens
token_count = len(tokens)
print(f"Token Count: {token_count}")
# Output (example): Token Count: 9

# 5. Decode tokens back to text (optional)
decoded_text = encoding.decode(tokens)
print(f"Decoded Text: {decoded_text}")
# Output (example): Decoded Text: Hello, world! How are you?
```
